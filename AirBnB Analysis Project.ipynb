## Vasulatization

#### 1 - What is this distribution of listing Price?

```python
plt.figure(figsize = (10,6))
sns.histplot(df['price'],bins = 50, kde = True , color = 'red')
plt.title("Distribution of listing Price")
plt.xlabel('Price $')
plt.ylabel('Frequancy')
plt.show()
```

#### 2 - How are different room distributed?

```python
df.head()
```

```python
df.columns
```

```python
df['room type']
```

```python
plt.figure(figsize = (8,5))
sns.countplot(x= 'room type', data = df, color = 'green')
plt.title("Room Type Distribution")
plt.xlabel("Room Type")
plt.ylabel("Count")
plt.show()
```

#### 3 - How are listing distributed across Neighbourhood?

```python
df.head(3)
```

```python
plt.figure(figsize = (12,8))
sns.countplot(y = 'neighbourhood group' , data = df, color = 'lightblue' , order=df['neighbourhood group'].value_counts().index)
plt.title("Number of Listing Distributed across Neighbourhood")
plt.xlabel("Counts")
plt.ylabel("Neighbourhood Group")
plt.show()
```

#### 4 - What is Relationship between Price and Room Type?

```python
plt.figure(figsize = (10,6))
sns.boxplot(x = 'room type', y = 'price' , hue = 'room type', data = df , palette = 'Set1')
plt.title("Price vs. Room Type")
plt.xlabel("Room Type")
plt.ylabel("Price ($)")
plt.legend(title = 'Room Type')
plt.show() 
```

#### 5 - how has the numbers of reviews change over with time?

```python
df['last review'] = pd.to_datetime(df['last review'])
reviews_over_time = df.groupby(df['last review'].dt.to_period('M')).size()


plt.figure(figsize=(12,6))
reviews_over_time.plot(kind='line', color ='red')
plt.title("Number of Reviews Over time")
plt.xlabel("Date")
plt.ylabel("Number of Reviews")
plt.show()
```

### Analysis Project - Summary

This Airbnb Analysis project explores key aspects of the Airbnb marketplace, including listing prices, room type distribution, neighborhood popularity, and customer review trends. The analysis provides valuable insights into accommodation patterns and customer engagement across different categories.

The price distribution analysis revealed that listings are spread across a wide range of price points, indicating diverse accommodation options for travelers with different budgets. Room type analysis showed that Entire Home/Apartments are the most common listings, while Hotel Rooms have the lowest representation in the dataset.

Neighborhood analysis highlighted Manhattan, Brooklyn, and Queens as the areas with the highest number of listings, making them the most active locations in the Airbnb market. Additionally, the relationship between price and room type demonstrated that Entire Homes/Apartments, Private Rooms, and Shared Rooms generally cover a broad price range, whereas Hotel Rooms tend to have a narrower but relatively higher pricing range.

Review trend analysis showed a noticeable increase in the number of reviews between 2019 and 2024 compared to 2014–2018, suggesting growing platform usage and increased customer engagement in recent years.

Overall, this project demonstrates how data analysis can be used to understand Airbnb market trends, customer preferences, pricing patterns, and listing distribution.
